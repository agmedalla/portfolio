## 1. Business Context & The Company Issue

RZR is a consumer debt-relief lead-generation agency. In short, we run online ads and partner with traffic networks to find people looking for debt relief, qualify their information, and sell those leads to debt-relief companies for a fixed price per lead (Cost-Per-Lead, or CPL).

At the time of this project, we were charging clients USD 30 per lead across 3,012 unique consumer leads, with only 4.4% turning into good quality leads. Leadership wanted to raise our price to USD 33 per lead (a 10% rate increase) to grow revenue. However, our clients pushed back and said they would only pay USD 33 if we proved we could improve our lead quality by 20% (bringing our baseline from 4.4% up to at least 5.3%).

**My Role:** Marketing Data Analyst

**Core Business Questions:**
1. How has our lead quality changed over time, and are we already trending toward that 20% improvement goal?
2. Which ad partners, ad creatives, and debt brackets are dragging down our 4.4% baseline—and which ones are actually performing well?
3. Which U.S. states bring in the most qualified leads, and where should we shift our ad budget so we can hit that 5.3% target and unlock the USD 33 price tier?

## 2. Technical Skills & Data Preparation Applied

To make sure our metrics were accurate before building the executive report, I used **Microsoft Excel** for initial data auditing and **Tableau Public** for data modeling, calculated fields, and interactive visualization:

- **Data Cleaning & Preparation (Microsoft Excel & Tableau Public):** I audited the raw 3,012-record lead log in Excel to check for missing values, standardize state abbreviations for geographic mapping, and verify debt-tier formatting (`USD 7,500` to `>USD 100,000`). In Tableau Public, I grouped fragmented raw `CallStatus` outcomes into clean binary tiers (`Good Quality` vs. `Bad Quality`) and built custom Tableau calculated fields—using `COUNTD([VendorLeadID])` to deduplicate repeat submissions, along with `COUNT`, `SUM`, and `IF/THEN` conditional logic—to compute `Quality Leads` and `Lead Quality Rate (%)`.
- **Data Visualization & Dashboard Engineering (Tableau Public):** I built a 5-sheet interactive dashboard combining a weekly time-series line chart with a linear trendline (`Apr 13` – `Sep 28`), side-by-side horizontal bar charts comparing volume against conversion rate, a color-encoded bar chart for debt tiers, and a U.S. choropleth map, complete with reference lines, floating legends, and cross-sheet action filters.

## 3. Data Visualization Walkthrough & Key Findings

### Q1 — Quality Trend Over Time (Weekly Line Chart with Trendline)
* **What It Shows:** Tracks weekly `Lead Quality Rate (%)` from mid-April through late September alongside a linear regression trendline.
* **What It Means:** Lead quality stagnated between **1% and 5%** from April through mid-July, keeping the cumulative baseline depressed at **4.4%**. However, starting in late August and September, conversion rates surged sharply—peaking above **13%**—proving that recent campaign adjustments have already demonstrated the capacity to surpass the **5.3% (+20% lift)** threshold required for the $33 CPL tier.

### Q2 — Partner Performance (Side-by-Side Volume vs. Quality Rate)
* **What It Shows:** Compares `Distinct count of VendorLeadID` (left) against `Lead Quality Rate (%)` (right) across all traffic partners.
* **What It Means:** **Google** accounts for **54% of total lead volume (~1,620 leads)** but converts at a dismal **3.0% quality rate**, acting as the primary anchor dragging down portfolio performance. Conversely, **AdKnowledge (6.4%)**, **Call Center (6.3%)**, and **Yahoo (5.5%)** convert at roughly double Google's efficiency, highlighting an immediate opportunity to improve portfolio quality simply by rebalancing partner spend.

### Q2 — Debt Level Performance (Color-Encoded Volume & Quality Bar Chart)
* **What It Shows:** Evaluates `Lead Quality Rate (%)` across 10 unsecured debt tiers (`$7,500` to `>$100,000`), with bar color intensity encoding lead volume.
* **What It Means:** Applicants in the lowest bracket (**$7,500–$10,000**) convert at barely **2.0%**, while ultra-high debt brackets (**$90,000–$100,000**) drop to **~1.1%**. The clear conversion "sweet spot" sits between **$10,001 and $70,000**, peaking at **6.2% quality** in both the **$20,001–$30,000** and **$50,001–$70,000** tiers.

### Q2 — Ad Creative Volume & Efficiency (Widget Comparison)

* **What It Means:** Several high-volume legacy widgets generate hundreds of clicks at sub-4% quality, whereas newer late-summer creative variants achieve **7% to 14%+ conversion rates** on smaller test budgets—representing high-upside creatives ready to be scaled.

### Q3 — Quality Leads by State (Choropleth Map)
* **What It Shows:** Maps total `Quality Leads` across the United States to identify geographic concentration.

## 4. Proposed Solutions & Business Impact

To bridge the gap from our **4.4% baseline** to the **5.3%+ quality benchmark (+20% lift)** and unlock the **$33 CPL** service rate, I delivered a four-pillar optimization roadmap:

1. **Reallocate Partner Spend:** Trim low-intent Google search traffic (3.0% quality) and shift budget toward high-converting partners—specifically **AdKnowledge (6.4%)**, **Call Center (6.3%)**, and **Yahoo (5.5%)**.
2. **Enforce Debt-Tier Filtering:** Tighten ad copy and landing-page pre-qualification filters to prioritize the **$10k–$70k unsecured debt sweet spot (up to 6.2% quality)** while suppressing sub-$10k leads (2.0%).
3. **Scale Winning September Ad Widgets:** Retire high-volume, low-converting legacy banners and reallocate impressions to September's top-performing `w-300250` creatives.
4. **Geo-Target High-Yield States:** Concentrate geo-bid modifiers on **CA, TX, and FL** while excluding zero-conversion states.

**How It Helped the Company:**  
This diagnostic equipped RZR's leadership with concrete, data-backed evidence to restructure their paid media acquisition strategy away from wasteful high-volume search traffic and gave the account team an actionable conversion-lift roadmap to justify raising client CPL pricing from **$30 to $33 per lead**.
