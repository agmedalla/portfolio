In this project, I analyzed international debt data sourced from The World Bank Organization. The dataset details the total amount of debt (in USD) owed by developing countries across various economic categories. Using PostgreSQL, I queried the database to answer three key questions:

- What is the number of distinct countries present in the database?
- What country has the highest amount of debt?
- What country has the lowest amount of repayments?

Below is a description of the table you will be working with:

## `international_debt` table

| Column | Definition | Data Type |
|-|-|-|
|country_name|Name of the country|`varchar`|
|country_code|Code representing the country|`varchar`|
|indicator_name|Description of the debt indicator|`varchar`|
|indicator_code|Code representing the debt indicator|`varchar`|
|debt|Value of the debt indicator for the given country (in current US dollars)|`float`|

I executed SQL queries to answer the three questions outlined in the project instructions.

```python
SELECT 
    COUNT(DISTINCT country_code) AS total_distinct_countries
FROM international_debt;
```

```python
SELECT country_name,
    SUM(debt) AS total_debt
FROM international_debt
GROUP BY country_name
ORDER BY total_debt DESC
LIMIT 1;
```

```python
SELECT country_name,  
	indicator_name, 
	debt AS lowest_repayment
FROM public.international_debt
	WHERE indicator_code = 'DT.AMT.DLXF.CD'
	ORDER BY debt ASC
	LIMIT 1;
```

### The Process & Analysis
For this project, I used descriptive analytics in SQL to summarize and explore international debt records from The World Bank. 

Because each country appeared multiple times across various financial categories, I started by deduplicating the records to establish the true geographic scope of the dataset. From there, I grouped and summed the financial figures by country to identify the largest overall borrower, and then filtered by a specific World Bank indicator code (`DT.AMT.DLXF.CD`) to isolate and rank principal debt repayments.

### The Results
My queries revealed that the dataset covers 124 distinct developing countries. When looking at total liabilities, China held the highest overall amount of debt at roughly 285.8 billion USD. 

On the other end of the spectrum, Timor-Leste recorded the lowest principal repayment on long-term external debt at 825,000 USD.
